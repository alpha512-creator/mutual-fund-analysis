```python
import os
import requests
import pandas as pd
```

```python
def fetch_fund_master():

    url = "https://api.mfapi.in/mf"
    print("Fetching global fund master list")

    try:
        response = requests.get(url)
        response.raise_for_status()

        master_list = response.json()

        df = pd.DataFrame(master_list)

        df.rename(columns={"schemeCode": "scheme_code", "schemeName": "scheme_name"}, inplace=True)

        output_path = "data/raw/fund_master.csv"
        os.makedirs(os.path.dirname(output_path), exist_ok=True)
        df.to_csv(output_path, index=False)

        print(f"Saved {len(df)} schemes to {output_path}")
        return df

    except requests.exceptions.RequestException as e:
        print(f"Failed to fetch master list: {e}")
        return None
```

```python
if __name__ == "__main__":
    fetch_fund_master()
```
