```python
import os
import glob
import pandas as pd
```

```python
def process_raw_nav_data(raw_dir="data/raw", processed_dir="data/processed"):
    print("STARTING DATA PROCESSING")

    os.makedirs(processed_dir, exist_ok=True)

    raw_files = glob.glob(os.path.join(raw_dir, "*.csv"))

    if not raw_files:
        print(f"No raw CSV files found in '{raw_dir}'.")
        return

    all_schemes_data = []

    for file_path in raw_files:
        file_name = os.path.basename(file_path)
        print(f"Processing: {file_name}...")

        df = pd.read_csv(file_path)

        df['date'] = pd.to_datetime(df['date'])

        df['nav'] = pd.to_numeric(df['nav'], errors='coerce')

        df = df.dropna(subset=['date', 'nav'])

        df = df.sort_values(by='date').reset_index(drop=True)

        processed_file_path = os.path.join(processed_dir, f"cleaned_{file_name}")
        df.to_csv(processed_file_path, index=False)

        all_schemes_data.append(df)
```

```python
if all_schemes_data:
    master_df = pd.concat(all_schemes_data, ignore_index=True)

    master_output_path = os.path.join(processed_dir, "master_nav_history.csv")
    master_df.to_csv(master_output_path, index=False)

    print("\nPROCESSING METRICS")
    print(f"Successfully processed {len(raw_files)} individual fund files.")
    print(f"Master dataset created at: {master_output_path}")
    print(f"Total records: {len(master_df)}")
    print("\nProcessed Data Sample:")
    print(master_df.head())
```

```python
if __name__ == "__main__":
    process_raw_nav_data()
```
